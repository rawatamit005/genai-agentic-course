```python
# 1. if / elif / else
marks = 72
if marks >= 90:
    print("A grade")
elif marks >= 60:
    print("B grade")
else:
    print("C grade")

# 2. and / or / not
age = 22
has_id = True
if age >= 18 and has_id:
    print("Entry mil sakti hai")

# 3. for loop
for i in range(5):
    print("Number:", i)

fruits = ["aam", "kela", "seb"]
for fruit in fruits:
    print(fruit.upper())

# 4. while loop
count = 0
while count < 3:
    print("count =", count)
    count += 1

# 5. Filter: sirf 2010 ke baad ki movies
films = [
    {"naam": "Dangal", "saal": 2016},
    {"naam": "3 Idiots", "saal": 2009},
    {"naam": "Lagaan", "saal": 2001},
    {"naam": "RRR", "saal": 2022},
]
nayi = []
for f in films:
    if f["saal"] > 2010:
        nayi.append(f["naam"])
print(nayi)
```
