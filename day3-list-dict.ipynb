```python
# 1. List
movies = ["Dangal", "3 Idiots", "RRR"]
print(movies[0], movies[-1])
print("-------------1------------")
movies.append("Lagaan")
movies.remove("RRR")
print(movies, len(movies))
print("-------------2------------")
print("Dangal" in movies)
print("-------------3------------")
for m in movies:
    print(m)
print("-------------4------------")
# 2. Dict
movie = {"naam": "Dangal", "saal": 2016, "rating": 8.3}
print(movie["naam"])
print("-------------5------------")
movie["hit"] = True
print(movie.get("director", "pata nahi"))
print("-------------6------------")
print(movie.keys())
print("-------------7------------")
print(movie.values())
print("-------------8------------")
for key, value in movie.items():
    print(key, "->", value)
print("-------------9------------")
# 3. List of dict (JSON jaisa)
films = [
    {"naam": "Dangal", "saal": 2016},
    {"naam": "3 Idiots", "saal": 2009},
    {"naam": "Lagaan", "saal": 2001},
]
for f in films:
    print(f"{f['naam']} ({f['saal']})")
print(films[1]["naam"])
```
